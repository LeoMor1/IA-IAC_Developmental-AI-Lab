## Define the Interaction class

Let's define an Interaction class that will be useful to intitialize the agent and to memorize the context

```python
class Interaction:
    """An interaction is a tuple (action, outcome) with a valence"""
    def __init__(self, action, outcome, valence):
        self.action = action
        self.outcome = outcome
        self.valence = valence

    def key(self):
        """ The key to find this interaction in the dictinary is the string '<action><outcome>'. """
        return f"{self.action}{self.outcome}"

    def __str__(self):
        """ Print interaction in the form '<action><outcome:<valence>' for debug."""
        return f"{self.action}{self.outcome}:{self.valence}"

    def __eq__(self, other):
        """ Interactions are equal if they have the same key """
        return self.key() == other.key()
```

## Define the Agent class

The agent is initialized with the list of interactions

The previous action and the predicted outcome are memorized in the attribute `_intended_interaction`.

```python
class Agent:
    """Creating our agent"""
    def __init__(self, _interactions):
        """ Initialize the dictionary of interactions"""
        self._interactions = {interaction.key(): interaction for interaction in _interactions}
        self._intended_interaction = self._interactions["00"]

    def action(self, _outcome):
        """ Tracing the previous cycle """
        previous_interaction = self._interactions[f"{self._intended_interaction.action}{_outcome}"]
        print(f"Action: {self._intended_interaction.action}, Prediction: {self._intended_interaction.outcome}, Outcome: {_outcome}, "
              f"Prediction: {self._intended_interaction.outcome == _outcome}, Valence: {previous_interaction.valence})")

        """ Computing the next interaction to try to enact """
        # TODO: Implement the agent's decision mechanism
        intended_action = 0
        # TODO: Implement the agent's prediction mechanism
        intended_outcome = 0
        # Memorize the intended interaction
        self._intended_interaction = self._interactions[f"{intended_action}{intended_outcome}"]
        return intended_action
```

## Environment1 class

```python
class Environment1:
    """ In Environment 1, action 0 yields outcome 0, action 1 yields outcome 1 """
    def outcome(self, _action):
        # return int(input("entre 0 1 ou 2"))
        if _action == 0:
            return 0
        else:
            return 1
```

## Environment2 class

```python
class Environment2:
    """ In Environment 2, action 0 yields outcome 1, action 1 yields outcome 0 """
    def outcome(self, _action):
        if _action == 0:
            return 1
        else:
            return 0
```

## Environment3 class

Environment 3 yields outcome 1 only when the agent alternates actions 0 and 1

```python
class Environment3:
    """ Environment 3 yields outcome 1 only when the agent alternates actions 0 and 1 """
    def __init__(self):
        """ Initializing Environment3 """
        self.previous_action = 0

    def outcome(self, _action):
        if _action == self.previous_action:
            _outcome = 0
        else:
            _outcome = 1
        self.previous_action = _action
        return _outcome
```

# ASSIGNMENT

Implement Agent4 that obtains positive valences in either Environment 1, 2, or 3.

Agent4 must be able to predict the outcome resulting from its next action depending on the context of the previous interaction.
Based on this prediction, it must select the action that will yield the heighest valence.

To do so, at the end of cycle `t`, Agent4 must memorize `interaction_t = (action_t, outcome_t)` that was just enacted.
The agent must choose the next `interaction_t+1` based on `interaction_t` (the context).
For each possible `action_t+1`, the agent must predict the expected `outcome_t+1`.
Based on this prediction, it must select the action that yields the highest `valence_t+1`.

## Create Agent4 by overriding the class Agent

You may add any attribute and method you deem usefull to the class Agent4

```python

class Agent4(Agent):
    def __init__(self, Interactions):
        super().__init__(Interactions)
        self.Memory = {}
        self.Context = None

    def action(self, _Outcome):
        Previous = self._intended_interaction
        super().action(_Outcome)
        Enacted = self._interactions[f"{Previous.action}{_Outcome}"]

        # Mémorise quelle interaction a suivi le contexte précédent
        if self.Context is not None:
            Transitions = self.Memory.setdefault(self.Context, {})
            Key = Enacted.key()
            Transitions[Key] = Transitions.get(Key, 0) + 1

        # On choisit la prochaine action selon ce qui a déjà suivi enacted
        NextContext = Enacted.key()
        Transitions = self.Memory.get(NextContext, {})

        Observed = [(self._interactions[Key], Count) for Key, Count in Transitions.items()]

        # Parmi les suites positives observées, garde la meilleure.
        Positive = [(Interaction, Count) for Interaction, Count in Observed if Interaction.valence > 0]

        if Positive:
            Chosen = None
            BestValence = None
            BestCount = -1

            for Interaction, Count in Positive:
                if (Chosen is None or Interaction.valence > BestValence or
                    (Interaction.valence == BestValence and Count > BestCount)
                ):
                    Chosen = Interaction
                    BestValence = Interaction.valence
                    BestCount = Count
        else:
            # Cherche une action qui n'a pas encore été testée dans ce contexte.
            TriedActions = {Interaction.action for Interaction, _ in Observed}
            Actions = list(dict.fromkeys(Interaction.action for Interaction in self._interactions.values()))
            UnexploredActions = [Action for Action in Actions if Action not in TriedActions]

            if UnexploredActions:
                ActionToTry = UnexploredActions[0]
                Predictions = [Interaction for Interaction in self._interactions.values() if Interaction.action == ActionToTry]

                Chosen = None
                BestValence = None

                for Interaction in Predictions:
                    if (Chosen is None or Interaction.valence > BestValence):
                        Chosen = Interaction
                        BestValence = Interaction.valence
            else:
                # Si toutes les actions ont été essayées, garde la meilleure
                # suite connue, en départageant par son nombre d'observations.
                Chosen = None
                BestValence = None
                BestCount = -1

                for Interaction, Count in Observed:
                    if (Chosen is None or Interaction.valence > BestValence or
                        (Interaction.valence == BestValence and Count > BestCount)
                    ):
                        Chosen = Interaction
                        BestValence = Interaction.valence
                        BestCount = Count


        self.Context = NextContext
        self._intended_interaction = Chosen
        return Chosen.action
```

```python
# @title Initialize the turtle environment

BORDER_WIDTH = 20

class ColabTurtleEnvironment:

    def __init__(self):
        """ Creating the Turtle window """
        bgcolor("lightGray")
        penup()
        goto(window_width() / 2, window_height()/2)
        face(0)
        pendown()
        color("green")

    def outcome(self, action):
        """ Enacting an action and returning the outcome """
        _outcome = 0
        for i in range(10):
            # _outcome = 0
            if action == 0:
                # move forward
                forward(10)
            elif action == 1:
                # rotate left
                left(4)
                forward(2)
            elif action == 2:
                # rotate right
                right(4)
                forward(2)

            # Bump on screen edge and return outcome 1
            if xcor() < BORDER_WIDTH:
                goto(BORDER_WIDTH, ycor())
                _outcome = 1
            if xcor() > window_width() - BORDER_WIDTH:
                goto(window_width() - BORDER_WIDTH, ycor())
                _outcome = 1
            if ycor() < BORDER_WIDTH:
                goto(xcor(), BORDER_WIDTH)
                _outcome = 1
            if ycor() > window_height() - BORDER_WIDTH:
                goto(xcor(), window_height() -BORDER_WIDTH)
                _outcome = 1

            # Change color
            if _outcome == 0:
                color("green")
            else:
                # Finit l'interaction
                color("red")
                # if action == 0:
                #     break
                if action == 1:
                    for j in range(10):
                        left(4)
                elif action == 2:
                    for j in range(10):
                        right(4)
                break

        return _outcome
```

```python
# @title Run the turtle environment
initializeTurtle()

# Parameterize the rendering
bgcolor("lightGray")
penup()
goto(window_width() / 2, window_height()/2)
face(0)
pendown()
color("green")
speed(10)

a = Agent4(interactions)
e = ColabTurtleEnvironment()

outcome = 0
for i in range(100):
    action = a.action(outcome)
    outcome = e.outcome(action)
```
